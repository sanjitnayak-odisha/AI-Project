```python
import cv2
import matplotlib.pyplot as plt

# Load uploaded images
img1 = cv2.imread("sample_image1.jpg")
img2 = cv2.imread("sample_image2.jpg")
```

```python

if img1 is None or img2 is None:
    print("Error: Image not found. Check file names.")
else:
    # Resize images to the same dimensions
    img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

    # Blend images
    blended = cv2.addWeighted(img1, 0.6, img2, 0.4, 0)

    # Save blended image (correct indentation)
    cv2.imwrite("blended_output.jpg", blended)

    # Display images
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
    plt.title("Original Image 1")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
    plt.title("Original Image 2")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(cv2.cvtColor(blended, cv2.COLOR_BGR2RGB))
    plt.title("Blended Image")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    print("Image blending completed successfully!")
```
