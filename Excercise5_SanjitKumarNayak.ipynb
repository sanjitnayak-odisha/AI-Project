```python
import gradio as gr
```

```python
#  Define chatbot response function
def chatbot_response(message, history):

    message = message.lower()

    if "hello" in message or "hi" in message:
        return "Hello! Welcome to my AI chatbot. How can I help you today?"

    elif "artificial intelligence" in message or "ai" in message:
        return (
            "Artificial Intelligence (AI) is a technology that enables "
            "machines to perform tasks that normally require human intelligence, "
            "such as learning, reasoning and problem-solving."
        )

    elif "cyber security" in message or "cybersecurity" in message:
        return (
            "Cybersecurity is the practice of protecting computers, networks "
            "and data from unauthorized access, cyberattacks and digital threats."
        )

    elif "python" in message:
        return (
            "Python is a popular, high-level programming language used in "
            "Artificial Intelligence, Data Science, Automation and Web Development."
        )

    elif "gradio" in message:
        return (
            "Gradio is a Python library used to create interactive web interfaces "
            "for machine learning models and AI applications."
        )

    else:
        return (
            "Thank you for your question! I am a demonstration chatbot. "
            "Please try asking about Artificial Intelligence, Python, "
            "Cybersecurity or Gradio."
        )
```

```python
#  Create Gradio ChatInterface
demo = gr.ChatInterface(
    fn=chatbot_response,
    title="My AI Conversational Chatbot",
    description=(
        "Welcome to my AI chatbot! "
        "Ask questions about Artificial Intelligence, Python, "
        "Cybersecurity and Gradio. "
        "You can also try the sample prompts below."
    ),
    examples=[
        "What is Artificial Intelligence?",
        "Explain Cybersecurity.",
        "What is Python programming?",
        "What is Gradio?"
    ]
)
```

```python
# Launch chatbot
demo.launch(share=True)
```
