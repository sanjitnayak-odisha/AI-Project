```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a DataFrame with random numerical values
np.random.seed(42)  # For reproducible results

df = pd.DataFrame({
    'f1': np.random.rand(10),
    'f2': np.random.rand(10),
    'f3': np.random.rand(10),
    'f4': np.random.rand(10)
})

# Display the DataFrame
print("Original DataFrame:")
print(df)

# 2. Scatter plot between f2 and f3
plt.scatter(df['f2'], df['f3'])

plt.xlabel('f2')
plt.ylabel('f3')
plt.title('Scatter Plot of f2 vs f3')
plt.grid(True)

plt.show()

# 3. Calculate row-wise arithmetic mean and store in f5
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

# Display the final DataFrame
print("\nDataFrame with f5 (Row-wise Mean):")
print(df)
```

```python
# 4. Export the complete DataFrame to an Excel file
df.to_excel("data_output.xlsx", index=False)

print("DataFrame exported successfully to data_output.xlsx")
```
